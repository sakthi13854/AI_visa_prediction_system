```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
```

```python
df = pd.read_csv(r'/home/sakthi/milestone2.csv')
```

```python
df.drop(columns=['SUBMISSION_DATE','DECISION_DATE'], inplace=True)
```

```python
encode_df = pd.get_dummies(
    df,
    columns=[
        'VISA_CLASS',
        'PROCESSING_CENTER',
        'EMPLOYER_STATE',
        'WORKSITE_STATE',
        'PROCESSING_CENTER',
        'SERVICE_OFFICE'
    ],
    drop_first=True
)
```

```python
encode_df.head()
```

```python
encode_df.drop(columns=[
    'APPLICATION_ID',
    'JOB_TITLE',
    'EMPLOYER_NAME',
    'SOC_TITLE'
], inplace=True)
```

```python
y = df['PROCESSING_DAYS']
```

```python
encode_df.drop(df['PROCESSING_DAYS'])
```

```python
features = [
'WAGE_RATE_OF_PAY_FROM',
'WAGE_RATE_OF_PAY_TO',
'PREVAILING_WAGE',
'PW_WAGE_LEVEL',
'NEW_EMPLOYMENT',
'CHANGE_EMPLOYER',
'AMENDED_PETITION',
 'APPLICATION_MONTH',
'SUBMISSION_YEAR',
'SUBMISSION_MONTH',
'SUBMISSION_DAY',
'SUBMISSION_QUARTER'
]
```

```python
x = encode_df.drop(columns=[
'APPLICATION_STATUS',
    "SOC_CODE",
    "WORKSITE_CITY",
    "PROCESSING_DAYS"

])
```

```python
print(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
def evaluate(y_true, y_pred):
    print("MAE:", mean_absolute_error(y_true, y_pred))
    print("RMSE:", np.sqrt(mean_squared_error(y_true, y_pred)))
    print("R2:", r2_score(y_true, y_pred))
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
evaluate(y_test,y_pred)
```

```python

rf = RandomForestRegressor(random_state=42)
rf.fit(x_train, y_train)

y_pred_rf = rf.predict(x_test)
```

```python


gb = GradientBoostingRegressor(random_state=42)
gb.fit(x_train, y_train)

y_pred_gb = gb.predict(x_test)
```

```python
evaluate(y_test, y_pred_rf)
```

```python
evaluate(y_test, y_pred_gb)
```

```python
results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest", "Gradient Boosting"],
    "MAE": [
        mean_absolute_error(y_test, y_pred),
        mean_absolute_error(y_test, y_pred_rf),
        mean_absolute_error(y_test, y_pred_gb)
    ],
    "RMSE": [
        np.sqrt(mean_squared_error(y_test, y_pred)),
        np.sqrt(mean_squared_error(y_test, y_pred_rf)),
        np.sqrt(mean_squared_error(y_test, y_pred_gb))
    ],
    "R2 Score": [
        r2_score(y_test, y_pred),
        r2_score(y_test, y_pred_rf),
        r2_score(y_test, y_pred_gb)
    ]
})
```

```python
results
```

### Since Random forest have higher R2 we choose Random Forest for hyper tuning

## hyper tuning

```python
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['sqrt', 'log2']
}
```

```python
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=3,
    scoring='r2',
    n_jobs=-1
)

grid_search.fit(x_train, y_train)
```

```python
print(grid_search.best_params_)
```

```python
best_rf = grid_search.best_estimator_
```

```python
y_pred_brf = best_rf.predict(x_test)
```

```python
evaluate(y_test,y_pred_brf)
```
